## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Create copies of the original datasets
stores_clean = stores.copy()
sales_clean = sales.copy()
features_clean = features.copy()

# Convert Date columns to datetime format
sales_clean['Date'] = pd.to_datetime(sales_clean['Date'], dayfirst=True)
features_clean['Date'] = pd.to_datetime(features_clean['Date'], dayfirst=True)

# Remove duplicate records
stores_clean = stores_clean.drop_duplicates()
sales_clean = sales_clean.drop_duplicates()
features_clean = features_clean.drop_duplicates()

# Check missing values
print("Missing values in Stores:")
print(stores_clean.isnull().sum())

print("\nMissing values in Sales:")
print(sales_clean.isnull().sum())

print("\nMissing values in Features:")
print(features_clean.isnull().sum())

# Fill missing markdown values with 0
markdown_cols = ['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5']

for col in markdown_cols:
    features_clean[col] = features_clean[col].fillna(0)

# Merge Sales with Stores
retail_data = sales_clean.merge(
    stores_clean,
    on='Store',
    how='left'
)

# Merge with Features using Store and Date
retail_data = retail_data.merge(
    features_clean,
    on=['Store', 'Date'],
    how='left',
    suffixes=('', '_feature')
)

# Sort data by Store, Department and Date
retail_data = retail_data.sort_values(
    by=['Store', 'Dept', 'Date']
).reset_index(drop=True)

# Display final dataset information
print("\nFinal Dataset Shape:", retail_data.shape)
print("\nFinal Dataset Columns:")
print(retail_data.columns.tolist())

retail_data.head()
````

### What all manipulations have you done and insights you found?

I first performed data cleaning and quality checks on the Stores, Sales, and Features datasets. I converted the date columns into datetime format, removed duplicates, handled missing values, and treated missing markdown values as zero. I then merged the datasets using Store and Date to create an integrated retail dataset with 421,570 records and 17 variables. Finally, I sorted the data by store, department, and date to prepare it for time-series analysis. From the initial analysis, I identified 45 stores, 3 store types, 81 departments, and significant variation in weekly sales, promotional markdowns, and economic factors such as fuel price, CPI, and unemployment. These variables will be further analyzed to understand sales patterns, detect anomalies, segment stores or customers, and build demand forecasting models."

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** There is no significant difference in average weekly sales between holiday and non-holiday weeks.

**Alternative Hypothesis (H₁):** There is a significant difference in average weekly sales between holiday and non-holiday weeks.


#### 2. Perform an appropriate statistical test.

````python
# Hypothesis 1 - Holiday vs Non-Holiday Sales

import pandas as pd
from scipy.stats import ttest_ind

# Load sales data
sales = pd.read_csv('/content/sales data-set (1).csv')

# Separate holiday and non-holiday sales
holiday_sales = sales[sales['IsHoliday'] == True]['Weekly_Sales']
non_holiday_sales = sales[sales['IsHoliday'] == False]['Weekly_Sales']

# Perform independent two-sample t-test
t_stat, p_value = ttest_ind(
    holiday_sales,
    non_holiday_sales,
    equal_var=False
)

print("Holiday Average Sales:", holiday_sales.mean())
print("Non-Holiday Average Sales:", non_holiday_sales.mean())
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject H0: There is a significant difference in average sales between holiday and non-holiday weeks.")
else:
    print("Fail to reject H0: There is no significant difference in average sales between holiday and non-holiday weeks.")
````

statistical Test: Independent Two-Sample T-Test.

````python
# Handling Missing Values & Missing Value Imputation

# Handling Missing Values & Missing Value Imputation

import pandas as pd

# Load datasets
sales = pd.read_csv('/content/sales data-set (1).csv')
stores = pd.read_csv('/content/stores data-set.csv')
features = pd.read_csv('/content/Features data set.csv')

# Convert Date columns
sales['Date'] = pd.to_datetime(sales['Date'], dayfirst=True)
features['Date'] = pd.to_datetime(features['Date'], dayfirst=True)

# Remove duplicate rows
sales = sales.drop_duplicates()
stores = stores.drop_duplicates()
features = features.drop_duplicates()

# Merge Sales + Stores + Features
retail_data = sales.merge(
    stores,
    on='Store',
    how='left'
)

retail_data = retail_data.merge(
    features,
    on=['Store', 'Date'],
    how='left',
    suffixes=('', '_feature')
)

# Check missing values before imputation
print("Missing values before imputation:")
print(retail_data.isnull().sum())

# Markdown columns
markdown_cols = [
    'MarkDown1',
    'MarkDown2',
    'MarkDown3',
    'MarkDown4',
    'MarkDown5'
]

# Impute Markdown missing values with 0
for col in markdown_cols:
    retail_data[col] = retail_data[col].fillna(0)

# Impute CPI and Unemployment using median
retail_data['CPI'] = retail_data['CPI'].fillna(
    retail_data['CPI'].median()
)

retail_data['Unemployment'] = retail_data['Unemployment'].fillna(
    retail_data['Unemployment'].median()
)

# Check missing values after imputation
print("\nMissing values after imputation:")
print(retail_data.isnull().sum())
````

Two imputation techniques were considered in the project:

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Handling Outliers & Outlier Treatments

import pandas as pd
import numpy as np

# Numerical columns for outlier detection
numeric_cols = [
    'Weekly_Sales',
    'Temperature',
    'Fuel_Price',
    'CPI',
    'Unemployment',
    'Size'
]

# Function to detect outliers using IQR
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return len(outliers), lower_bound, upper_bound

# Detect outliers
print("Outlier Detection using IQR:\n")

for col in numeric_cols:
    count, lower, upper = detect_outliers_iqr(retail_data, col)

    print(f"{col}:")
    print(f"  Outliers: {count}")
    print(f"  Lower Bound: {lower:.2f}")
    print(f"  Upper Bound: {upper:.2f}")
    print()
````

````python
# Outlier Treatment using IQR Capping

outlier_cols = [
    'Weekly_Sales',
    'Temperature',
    'Unemployment'
]

for col in outlier_cols:
    Q1 = retail_data[col].quantile(0.25)
    Q3 = retail_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Cap extreme values
    retail_data[col] = retail_data[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed using IQR-based capping.")
````

````python
# Check outliers after treatment

print("Outliers after treatment:\n")

for col in outlier_cols:
    Q1 = retail_data[col].quantile(0.25)
    Q3 = retail_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    count = (
        (retail_data[col] < lower_bound) |
        (retail_data[col] > upper_bound)
    ).sum()

    print(f"{col}: {count}")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

### Outlier Treatment Conclusion

Outliers were detected using the IQR method and treated using IQR-based capping. The extreme values in `Weekly_Sales`, `Temperature`, and `Unemployment` were capped at their respective IQR boundaries instead of removing the observations.

After treatment, the outlier check showed **0 outliers** in all three treated variables. This confirms that the outlier treatment was successfully applied while preserving the original dataset records.

### ML Model - 3

````python
# ML Model - 3 Implementation
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create Model 3
gb_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# Fit the model
gb_model.fit(X_train, y_train)

# Predict
y_pred_gb = gb_model.predict(X_test)

# Evaluation
mae_gb = mean_absolute_error(y_test, y_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))
r2_gb = r2_score(y_test, y_pred_gb)

print("Gradient Boosting Regression - Model 3")
print("--------------------------------------")
print("MAE :", mae_gb)
print("RMSE:", rmse_gb)
print("R²  :", r2_gb)
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Visualizing evaluation Metric Score chart
import matplotlib.pyplot as plt

metrics = ['MAE', 'RMSE', 'R²']
scores = [mae_gb, rmse_gb, r2_gb]

plt.figure(figsize=(8,5))
plt.bar(metrics, scores)
plt.title('Gradient Boosting Regression - Evaluation Metrics')
plt.xlabel('Metrics')
plt.ylabel('Score')
plt.show()
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 3 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import RandomizedSearchCV

# Create Gradient Boosting model
gb = GradientBoostingRegressor(
    random_state=42
)

# Hyperparameter search space
param_grid_gb = {
    'n_estimators': [50, 100, 150],
    'learning_rate': [0.05, 0.1, 0.15],
    'max_depth': [2, 3, 4],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Randomized Search CV
random_search_gb = RandomizedSearchCV(
    estimator=gb,
    param_distributions=param_grid_gb,
    n_iter=5,
    cv=3,
    scoring='neg_mean_squared_error',
    random_state=42,
    n_jobs=-1,
    verbose=2
)

# Fit the algorithm
print("Starting Hyperparameter Optimization...")
random_search_gb.fit(X_train, y_train)

# Best parameters
print("\nBest Parameters:")
print(random_search_gb.best_params_)

# Get optimized model
best_gb_model = random_search_gb.best_estimator_

# Predict on the test data
y_pred_gb_tuned = best_gb_model.predict(X_test)

print("\nPrediction completed successfully.")
print("First 10 predictions:")
print(y_pred_gb_tuned[:10])
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Evaluation of optimized Model 3
mae_gb_tuned = mean_absolute_error(y_test, y_pred_gb_tuned)
rmse_gb_tuned = np.sqrt(mean_squared_error(y_test, y_pred_gb_tuned))
r2_gb_tuned = r2_score(y_test, y_pred_gb_tuned)

print("Optimized Gradient Boosting Regression - Model 3")
print("------------------------------------------------")
print("MAE :", mae_gb_tuned)
print("RMSE:", rmse_gb_tuned)
print("R²  :", r2_gb_tuned)
````
